Desafios para fixação

 Desafio 1: Listar animais não adotados

```python
def animais_nao_adotados():
  resposta = supabase.table("animais").select("*").eq("status", "Disponível").execute()
  return resposta.data
```

Desafio 2: Buscar por espécie


```python
def buscar_por_especie(especie):
  resposta = supabase.table("animais").select("*").eq("especie", especie).execute()
  return resposta.data
```

Desafio 3: Relatório ordenado

```python
def relatorio_ordenado():
  resposta = supabase.table("animais").select("*").order("idade", desc=False).execute()
  animais = resposta.data
  for animal in animais:
    print(f"Nome: {animal.get('nome', 'N/A')}, Idade: {animal.get('idade', 'N/A')}")
```

 Desafio 4: Interface simples com entrada de dados

```python
def inserir_animal():
  nome = input("Digite o nome do animal: ")
  especie = input("Digite a espécie do animal: ")
  raca = input("Digite a raça do animal: ")
  porte = input("Digite o porte do animal: ")
  sexo = input("Digite o sexo do animal (M/F): ")
  idade = int(input("Digite a idade do animal: "))
  cor = input("Digite a cor do animal: ")
  status = input("Digite o status do animal (Disponível/Adotado): ")

  novo_animal = {
      'nome': nome,
      'especie': especie,
      'raca': raca,
      'porte': porte,
      'sexo': sexo,
      'idade': idade,
      'cor': cor,
      'dataentrada': None,
      'status': status
  }

  resposta = supabase.table("animais").insert(novo_animal).execute()
  print(resposta.data)

inserir_animal()
```
